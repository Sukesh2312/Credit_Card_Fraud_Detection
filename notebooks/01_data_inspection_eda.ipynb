```python
def fraud_statistics(df, year):

    total = len(df)
    fraud_count = (df["Class"] == 1).sum()
    legitimate_count = (df["Class"] == 0).sum()

    fraud_percentage = fraud_count / total * 100
    legitimate_percentage = legitimate_count / total * 100

    return {
        "Dataset": year,
        "Total Transactions": total,
        "Legitimate": legitimate_count,
        "Fraud": fraud_count,
        "Legitimate %": legitimate_percentage,
        "Fraud %": fraud_percentage
    }


comparison = pd.DataFrame([
    fraud_statistics(df_2013, "2013"),
    fraud_statistics(df_2023, "2023")
])

display(comparison)
```

```python
fraud_2013 = comparison.loc[
    comparison["Dataset"] == "2013", "Fraud %"
].iloc[0]

fraud_2023 = comparison.loc[
    comparison["Dataset"] == "2023", "Fraud %"
].iloc[0]

print(f"2013 fraud percentage: {fraud_2013:.4f}%")
print(f"2023 fraud percentage: {fraud_2023:.4f}%")

if fraud_2023 > fraud_2013:
    print("\n2023 has a higher fraud percentage than 2013.")
    print("Therefore, 2023 is less imbalanced than 2013.")

elif fraud_2023 < fraud_2013:
    print("\n2023 has a lower fraud percentage than 2013.")
    print("Therefore, 2023 is more imbalanced than 2013.")

else:
    print("\nBoth datasets have the same fraud percentage.")
```

```python
def split_features_target(df):

    X = df.drop(columns=["Class"])
    y = df["Class"]

    return X, y
```

```python
X_2013, y_2013 = split_features_target(df_2013)
X_2023, y_2023 = split_features_target(df_2023)

print("2013 X shape:", X_2013.shape)
print("2013 y shape:", y_2013.shape)

print("\n2023 X shape:", X_2023.shape)
print("2023 y shape:", y_2023.shape)
```

```python
X_train_2013, X_test_2013, y_train_2013, y_test_2013 = train_test_split(
    X_2013,
    y_2013,
    test_size=0.20,
    random_state=42,
    stratify=y_2013
)

X_train_2023, X_test_2023, y_train_2023, y_test_2023 = train_test_split(
    X_2023,
    y_2023,
    test_size=0.20,
    random_state=42,
    stratify=y_2023
)

print("2013:")
print("Training:", X_train_2013.shape)
print("Testing :", X_test_2013.shape)

print("\n2023:")
print("Training:", X_train_2023.shape)
print("Testing :", X_test_2023.shape)
```

```python
def show_class_distribution(y, name):

    counts = y.value_counts().sort_index()
    percentages = y.value_counts(normalize=True).sort_index() * 100

    result = pd.DataFrame({
        "Count": counts,
        "Percentage": percentages.round(4)
    })

    result.index = ["Legitimate", "Fraud"]

    print(f"\n{name}")
    display(result)
```

```python
show_class_distribution(y_2013, "2013 Original")
show_class_distribution(y_train_2013, "2013 Training")
show_class_distribution(y_test_2013, "2013 Testing")

show_class_distribution(y_2023, "2023 Original")
show_class_distribution(y_train_2023, "2023 Training")
show_class_distribution(y_test_2023, "2023 Testing")
```

```python
def get_scaling_columns(X):

    possible_columns = ["Time", "Amount"]

    return [
        column
        for column in possible_columns
        if column in X.columns
    ]


scale_2013 = get_scaling_columns(X_train_2013)
scale_2023 = get_scaling_columns(X_train_2023)

print("2013 columns to scale:", scale_2013)
print("2023 columns to scale:", scale_2023)
```

```python

def get_feature_names_from_pipeline(pipeline):

    preprocessor = pipeline.named_steps["preprocessing"]

    feature_names = preprocessor.get_feature_names_out()

    # Remove transformer prefixes
    feature_names = [
        name.split("__")[-1]
        for name in feature_names
    ]

    return feature_names
```

```python
preprocessor_2013 = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), scale_2013)
    ],
    remainder='passthrough'
)

rf_2013 = Pipeline(steps=[
    ('preprocessing', preprocessor_2013),
    ('classifier', RandomForestClassifier(random_state=RANDOM_STATE))
])

# Fit the pipeline to the training data for 2013
rf_2013.fit(X_train_2013, y_train_2013)

print("2013 pipeline created and fitted.")
```

```python
preprocessor_2023 = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), scale_2023)
    ],
    remainder='passthrough'
)

rf_2023 = Pipeline(steps=[
    ('preprocessing', preprocessor_2023),
    ('classifier', RandomForestClassifier(random_state=RANDOM_STATE))
])

# Fit the pipeline to the training data for 2023
rf_2023.fit(X_train_2023, y_train_2023)

print("2023 pipeline created and fitted.")
```

```python
features_2013 = get_feature_names_from_pipeline(rf_2013)
features_2023 = get_feature_names_from_pipeline(rf_2023)

print("2013 number of features:", len(features_2013))
print("2023 number of features:", len(features_2023))

print("\n2013 features:")
print(features_2013)

print("\n2023 features:")
print(features_2023)
```

```python
def get_feature_importance(model_pipeline):

    model = model_pipeline.named_steps["model"]

    feature_names = get_feature_names_from_pipeline(model_pipeline)

    importance = model.feature_importances_

    result = pd.DataFrame({
        "Feature": feature_names,
        "Importance": importance
    })

    return result.sort_values(
        by="Importance",
        ascending=False
    )
```

```python
def get_feature_importance(model_pipeline):

    # The original function had 'model' but the pipeline step is named 'classifier'
    model = model_pipeline.named_steps["classifier"]

    feature_names = get_feature_names_from_pipeline(model_pipeline)

    importance = model.feature_importances_

    result = pd.DataFrame({
        "Feature": feature_names,
        "Importance": importance
    })

    return result.sort_values(
        by="Importance",
        ascending=False
    )

rf_importance_2013 = get_feature_importance(rf_2013)

display(rf_importance_2013.head(15))
```

```python
def plot_feature_importance(
    importance_df,
    dataset_name,
    model_name="Random Forest",
    top_n=15
):

    top_features = importance_df.head(top_n).sort_values(
        by="Importance"
    )

    plt.figure(figsize=(10, 7))

    plt.barh(
        top_features["Feature"],
        top_features["Importance"]
    )

    plt.xlabel("Feature Importance")
    plt.ylabel("Feature")
    plt.title(
        f"Top {top_n} Features — {model_name} — {dataset_name}"
    )

    plt.tight_layout()
    plt.show()
```

```python
def evaluate_model(model_pipeline, X_test, y_test, model_name, dataset_name):
    y_pred = model_pipeline.predict(X_test)
    y_pred_proba = model_pipeline.predict_proba(X_test)[:, 1]

    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_pred_proba)
    pr_auc = average_precision_score(y_test, y_pred_proba)

    return {
        "Dataset": dataset_name,
        "Model": model_name,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
    }


all_results = []

# Evaluate Random Forest for 2013
all_results.append(evaluate_model(rf_2013, X_test_2013, y_test_2013, "Random Forest", "2013"))

# Evaluate Random Forest for 2023
all_results.append(evaluate_model(rf_2023, X_test_2023, y_test_2023, "Random Forest", "2023"))

# Evaluate XGBoost for 2013
all_results.append(evaluate_model(xgb_2013, X_test_2013, y_test_2013, "XGBoost", "2013"))

# Evaluate XGBoost for 2023
all_results.append(evaluate_model(xgb_2023, X_test_2023, y_test_2023, "XGBoost", "2023"))

results_df = pd.DataFrame(all_results)


for metric in [
    "Precision",
    "Recall",
    "F1",
    "ROC-AUC",
    "PR-AUC"
]:

    print("\n" + "=" * 50)
    print(metric)
    print("=" * 50)

    display(
        results_df[
            ["Dataset", "Model", metric]
        ].sort_values(
            by=metric,
            ascending=False
        )
    )
```
